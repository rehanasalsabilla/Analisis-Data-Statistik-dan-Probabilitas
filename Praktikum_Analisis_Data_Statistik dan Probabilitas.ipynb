Untuk menyelesaikan soal 1A dan 1B, menggunakan **Teorema Bayes**. Teorema Bayes digunakan untuk menghitung probabilitas posterior dari suatu kejadian berdasarkan informasi sebelumnya.

## ***Soal 1A : Peluang Mengidap Flu Setelah Divaksinasi***<br>
Untuk menghitung peluang seseorang mengidap flu setelah divaksinasi, langkah-langkahnya adalah sebagai berikut:

**Langkah-langkah:**
1. Menghitung jumlah orang yang telah divaksin ($n_{\text{vaksin}}$).
2. Menghitung jumlah orang yang mengalami flu dan telah divaksin ($n_{\text{flu dan vaksin}}$).
3. Menggunakan teorema Bayes untuk menghitung peluang seseorang akan mengidap flu apabila telah divaksin.

**Penyelesaian:**
- $n_{\text{vaksin}} = 200 + 4000 = 4200$
- $n_{\text{flu dan vaksin}} = 200$

Gunakan teorema Bayes:
$$
P(\text{flu} \ | \ \text{vaksin}) = \frac{n_{\text{flu dan vaksin}}}{n_{\text{vaksin}}}
$$

**Perhitungan:**
$$
\begin{align*}
P(\text{flu} \ | \ \text{vaksin}) &= \frac{n(\text{flu} \cap \text{vaksin})}{n(\text{vaksin})} \\
&= \frac{200}{4200} \\
&\approx 0.0476
\end{align*}
$$
<br>
## ***Soal 1B: Peluang Mengidap Flu Tanpa Vaksinasi***
Untuk menghitung peluang seseorang mengidap flu tanpa vaksinasi, langkah-langkahnya adalah sebagai berikut:

**Langkah-langkah:**
1. Menghitung jumlah orang yang belum divaksin ($n_{\text{tidak vaksin}}$).
2. Menghitung jumlah orang yang mengalami flu dan belum divaksin ($n_{\text{flu dan tidak vaksin}}$).
3. Menggunakan teorema Bayes untuk menghitung peluang seseorang akan mengidap flu apabila belum divaksin.

**Penyelesaian:**
- $n_{\text{tidak vaksin}} = 900 + 9000 = 9900$
- $n_{\text{flu dan tidak vaksin}} = 900$

Gunakan teorema Bayes:
$$
P(\text{flu} \ | \ \text{tidak vaksin}) = \frac{n_{\text{flu dan tidak vaksin}}}{n_{\text{tidak vaksin}}}
$$

**Perhitungan:**
$$
\begin{align*}
P(\text{flu} \ | \ \text{tidak vaksin}) &= \frac{n(\text{flu} \cap \text{tidak vaksin})}{n(\text{tidak vaksin})} \\
&= \frac{900}{9900} \\
&\approx 0.0909
\end{align*}
$$
<br>
Jadi, peluang seseorang akan mengidap flu apabila telah divaksin ($P(\text{flu} \ | \ \text{vaksin})$) adalah 0.0476, sedangkan peluang seseorang akan mengidap flu apabila belum divaksin ($P(\text{flu} \ | \ \text{tidak vaksin})$) adalah 0.0909.

### A. Jika seorang pekerja dipilih secara acak dan ternyata memiliki pengalaman kerja kurang dari 3 tahun, berapa probabilitas bahwa pekerja tersebut berasal dari Divisi Pengembangan?

```python
# Mengatur probabilitas bahwa seorang pekerja memiliki pengalaman kerja kurang dari 3 tahun untuk setiap divisi
prob_pengembangan_min3 = 0.3
prob_pemasaran_min3 = 0.7

# Mengatur probabilitas bahwa seorang pekerja berasal dari masing-masing divisi
prob_pengembangan = 0.5
prob_pemasaran = 0.5

# Menghitung probabilitas bahwa seorang pekerja berasal dari setiap divisi dan memiliki pengalaman kurang dari 3 tahun
prob_pengembangan_dan_min3 = prob_pengembangan * prob_pengembangan_min3
prob_pemasaran_dan_min3 = prob_pemasaran * prob_pemasaran_min3
prob_min3 = prob_pengembangan_dan_min3 + prob_pemasaran_dan_min3

# Menghitung probabilitas bahwa seorang pekerja dipilih secara acak dari Divisi Pengembangan jika memiliki pengalaman kurang dari 3 tahun
prob_pengembangan_acak = prob_pengembangan_dan_min3 / prob_min3

Soal_3_A = prob_pengembangan_acak
print(f"3A. Jawaban : {Soal_3_A:.2f}")
```

### B. Rasio probabilitas pekerja berasal dari Divisi Pengembangan terhadap Divisi Pemasaran

```python
Soal_3_B = prob_pengembangan / prob_pemasaran
print(f"3B. Jawaban : {Soal_3_B:.2f}")
```

**Penjelasan Penyelesaian Soal 3** <br>
Dalam menyelesaikan soal ini, ada beberapa konsep dasar probabilitas yang digunakan:

**Aturan Probabilitas Total:** Digunakan untuk menghitung probabilitas suatu kejadian dengan mempertimbangkan semua kemungkinan kondisi atau kategori yang mungkin. Aturan probabilitas total digunakan untuk menghitung probabilitas bahwa seorang pekerja yang dipilih secara acak berasal dari Divisi Pengembangan, jika diketahui bahwa pekerja tersebut memiliki pengalaman kerja kurang dari 3 tahun.

**Rasio Probabilitas:** Untuk menghitung rasio antara probabilitas bahwa seorang pekerja berasal dari Divisi Pengembangan dibandingkan dengan probabilitas bahwa pekerja tersebut berasal dari Divisi Pemasaran, jika diketahui bahwa pekerja tersebut memiliki pengalaman kerja kurang dari 3 tahun.

### **Langkah Penyelesaian :**

#### ***Soal 3.A: Probabilitas bahwa pekerja tersebut berasal dari Divisi Pengembangan jika memiliki pengalaman kurang dari 3 tahun***

Gunakan aturan probabilitas total:

$$
P(\text{Pengembangan | kurang dari 3 tahun}) = \frac{P(\text{Pengembangan}) \times P(\text{kurang dari 3 tahun | Pengembangan})}{P(\text{kurang dari 3 tahun})}
$$
<br>
Ketika sudah memiliki nilai-nilai probabilitas yang diperlukan:
<br>

$$
\begin{align*}
P(\text{Pengembangan}) &= 0.5 \\
P(\text{kurang dari 3 tahun | Pengembangan}) &= 0.3 \\
P(\text{kurang dari 3 tahun}) &= 0.15 + 0.35 = 0.50 \quad \text{(Total probabilitas pekerja yang memiliki pengalaman kurang dari 3 tahun)}
\end{align*}
$$

Gunakan rumus di atas untuk menghitung probabilitasnya:

$$
\begin{align*}
P(\text{Pengembangan | kurang dari 3 tahun}) &= \frac{0.5 \times 0.3}{0.50} \\
&= \frac{0.15}{0.50} \\
&= 0.30
\end{align*}
$$
Jadi, hasilnya adalah 0.30.
<br>

#### **Soal 3.B: Rasio probabilitas pekerja berasal dari Divisi Pengembangan terhadap Divisi Pemasaran**
Dari soal, diketahui bahwa probabilitas pekerja berasal dari Divisi Pengembangan adalah 0,30 atau 30%. Kemudian, karena total probabilitas harus sama dengan 1 (atau 100%), maka probabilitas pekerja berasal dari Divisi Pemasaran dapat dihitung dengan mengurangkan probabilitas pekerja berasal dari Divisi Pengembangan dari 1.

Jadi, jika probabilitas pekerja berasal dari Divisi Pengembangan adalah 0,30, maka probabilitas pekerja berasal dari Divisi Pemasaran dapat dihitung dengan rumus sebagai berikut.

**Probabilitas pekerja divisi pengembangan :**

$$
P(\text{Divisi M}) = 1 - P(\text{Divisi P}) = 1 - 0,30 = 0,70
$$
<br>

Hitung rasio antara probabilitas pekerja berasal dari Divisi Pengembangan<br>
**Rasio Probabilitas**:
   $$
   \text{Rasio Probabilitas} = \cfrac{P(\text{Pengembangan})}{P(\text{Pemasaran})}
   \cfrac{P(\text{Divisi P})}{P(\text{Divisi M})} = \cfrac{0.30}{0.70} = 0.43
   $$
<br>
Jadi, Ini menunjukkan bahwa kemungkinan lebih tinggi untuk memilih pekerja dengan pengalaman kurang dari 3 tahun dari Divisi Pemasaran daripada dari Divisi Pengembangan.

### **Soal 5**: Berapa probabilitas sebuah kemasan perlu diganti, asumsikan bahwa probabilitas sebuah sekrup cacat adalah 0.01 dan setiap kemasan berisi 10 sekrup?

Clue: $P(X>1)$ bisa dicari dengan
$$P(X>1) = 1 - P(X=0) - P(X=1)$$

```python
jumlah_sekrup_per_kemasan = 10  # Setiap kemasan berisi 10 sekrup
probabilitas_cacat = 0.01  # Probabilitas sebuah sekrup cacat

# Menghitung probabilitas setidaknya satu sekrup cacat dalam satu kemasan
prob_setidaknya_satu_cacat = 1 - binom.pmf(k=0, n=jumlah_sekrup_per_kemasan, p=probabilitas_cacat)

# Menampilkan hasil
Soal_5 = prob_setidaknya_satu_cacat
print(Soal_5)
```

**Penjelasan Penyelesaian Soal 5** <br>
Langkah-langkah dalam menyelesaikan soal ini:

1. Disini memiliki probabilitas bahwa satu sekrup adalah cacat (probabilitas_cacat).
2. Mencari probabilitas bahwa setidaknya satu dari sepuluh sekrup dalam satu kemasan adalah cacat.
3. Gunakan rumus distribusi binomial kumulatif: 1 - binom.pmf(k=0, n=jumlah_sekrup_per_kemasan, p=probabilitas_cacat). <br>
Di mana:
  - k adalah 0 karena kita ingin setidaknya satu sekrup cacat.
  - n adalah jumlah percobaan (jumlah sekrup per kemasan).
  - p adalah probabilitas keberhasilan dalam satu percobaan (probabilitas sebuah sekrup cacat).

##### **Menggunakan rumus distribusi binomial kumulatif:**
$$P(X \geq 1) = 1 - P(X=0)$$

Di mana $P(X=0)$ adalah probabilitas bahwa tidak ada sekrup cacat dalam satu kemasan.

Untuk menghitung $P(X=0)$, kita menggunakan fungsi probabilitas massa binomial (binom.pmf) dengan
  - $k=0$ (tidak ada sekrup cacat),
  - $n=10$ (jumlah sekrup per kemasan),
  - $p=0.01$ (probabilitas sekrup cacat).
  
Jadi:

$$P(X=0) = \binom{10}{0} \times (0.01)^0 \times (1-0.01)^{10-0}$$
$$P(X=0) = 1 \times 1 \times 0.99^{10}$$
$$P(X=0) = 0.99^{10}$$

<br>
Sekarang, gunakan hasil ini untuk mencari $P(X \geq 1)$:

$$P(X \geq 1) = 1 - 0.99^{10}$$
<br>
Nilai dari $0.99^{10}$, kemudian kurangkan dari 1 untuk mendapatkan probabilitas bahwa setidaknya satu sekrup dalam satu kemasan adalah cacat.

$$P(X \geq 1) = 1 - 0.99^{10} = 1 - 0.9043820750088044 \approx 0.09561792499119592$$
<br>
Jadi, probabilitas bahwa sebuah kemasan perlu diganti adalah sekitar 0.0956 atau sekitar 9.56%.

Hitung <br>

```python
# Solver soal 6
import numpy as np

# Data skor teknikal dan skor artistik
skor_teknik = np.array([96, 88, 92, 90, 94, 89, 93, 87, 95, 91, 90, 92, 89, 96, 88, 92, 89, 87, 96, 88])
skor_artistik = np.array([91, 87, 89, 94, 90, 85, 88, 92, 93, 86, 91, 88, 90, 95, 89, 91, 86, 90, 95, 89])

# A. Expectation (Nilai Harapan)
expectation_teknik = np.mean(skor_teknik)
expectation_artistik = np.mean(skor_artistik)


# B. Variance (Varians)
variance_teknik = sum((x - expectation_teknik)** 2 for x in skor_teknik ) / len(skor_teknik)
variance_artistik = sum((y - expectation_artistik)** 2 for y in skor_artistik ) / len(skor_artistik)


# C. Covariance Matrix (Matriks Kovarians)
covariance_matrix = np.cov(skor_teknik, skor_artistik)

# Menampilkan hasil
print("A. Expectation (Nilai Harapan):")
print("Nilai Harapan Skor Teknik:", expectation_teknik)
print("Nilai Harapan Skor Artistik:", expectation_artistik)
print("\nB. Variance (Varians):")
print("Varians Skor Teknik:", variance_teknik)
print("Varians Skor Artistik:", variance_artistik)
print("\nC. Covariance Matrix (Matriks Kovarians):")
print(covariance_matrix)
```

### **C. Perhitungan matriks kovarians:**

Matriks kovarians mengukur seberapa dua variabel berubah bersamaan. Rumusnya adalah Σ((X - μx) * (Y - μy)) / (n - 1).
Dalam case ini, menggunakan np.cov() untuk menghitung kovarians antara skor teknikal dan skor artistik.
